### Analysis

```python
network_configs = network_settings.networks_types
neighbour_influences = dgp_settings.n_influence_list
assignment_type = "random"
lr_estimator = LinearRegressionEstimator()
network_extractor = NetworkFeatureExtractor()

all_results_df = pd.DataFrame()
for network_type, p_edges in tqdm(network_configs):
    for influence in neighbour_influences:
        file_path = f"/Users/polinarevina/Desktop/thesis_new/src/simulation_results/sim_{assignment_type}_{network_type}_influence{influence}.pkl"
        
        with open(file_path, "rb") as f:
            data = pickle.load(f)
        
        results_dict = defaultdict(lambda: {"pvalue": [], "coef": []})

        for i_sim in range(dgp_settings.n_sim):

            adj_matrix = data[i_sim]["adj_matrix"]
            individ_data_sim = data[i_sim]["individ_data"]
            treatment_assignment = individ_data_sim["group"]
            df_features = network_extractor.count_treated_neighbors(
                adj_matrix=adj_matrix,
                treatment_assignment=treatment_assignment
            )
            individ_data_sim = individ_data_sim.merge(df_features)

            for feature in  [["treated_neighbors_exposure"], ["treated_neighbors_count"]]:
                coef, pvalue = lr_estimator.calculate_results(
                        individ_data=individ_data_sim,
                        additional_feature_names=feature
                    )
                results_dict[feature[0]]["pvalue"].append(pvalue)
                results_dict[feature[0]]["coef"].append(coef)
    
        for feature in [["treated_neighbors_exposure"], ["treated_neighbors_count"]]:
            results_plotter = ResultsPlotter(
                        estimated_effect=results_dict[feature[0]]["pvalue"],
                        true_effect=dgp_settings.treatment_effect_mean,
                        network_type=network_type,
                        assignment_type=assignment_type,
                        neighbour_influence=influence,
                        treatment_mean=dgp_settings.treatment_effect_mean,
                        additional_feature_names=feature[0]
            )
            results_plotter.calculate_results()
            results_df = results_plotter.create_results()
            all_results_df = pd.concat([all_results_df, results_df], ignore_index=True)
```

Naiv estimations that don't account for the presence of network effect have on average 107% bias. Different network significantly impact the bias, with Erdos-Renyi 96%, Barabasi-Albert 126% and Watts-Strogatz 100%. With the increase of neigbour influence the bias is also increasing. 

The inclusion of network properties such as degree centrality, betweenness centrality, and community structure appears to influence the bias, leading to 16% bias for Barabasi-Albert graph and 5% for other graph models. On the other hand, inclusion of just community feature decreased only bias in Watts-Strogatz model and for other network type even increased the bias. Adjustments using vector representation of nodes that were obtained using node2vec also decreases bias to around 42-46% for every network type.
